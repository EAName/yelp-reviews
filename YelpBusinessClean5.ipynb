# Clean BusinessParking

```python
dfbp = df.copy(deep = True)
```

```python
#Create business parking df to run methods to create separate columns from business parking

dfbp = dfbp[['business_id', 'name', 'market', 'BusinessParking']]
```

Need to convert the strings in businessparking to dicts with import_eval, fill NA with {},
then use json normalize to create separate columns

```python
# fill bp nan with {}
dfbp.BusinessParking = dfbp.BusinessParking.fillna('{}') 
```

```python
dfbp.info()
```

```python
from ast import literal_eval

dfbp.BusinessParking = dfbp.BusinessParking.apply(literal_eval)
```

```python
dfbp2 = dfbp.join(pd.json_normalize(dfbp['BusinessParking']))

dfbp2.head(10)
```

```python
#Convert True/False to 1,0 for the new columns

dfbp2.garage = dfbp2.garage.replace({True: 1, False: 0})
dfbp2.street = dfbp2.street.replace({True: 1, False: 0})
dfbp2.validated = dfbp2.validated.replace({True: 1, False: 0})
dfbp2.lot = dfbp2.lot.replace({True: 1, False: 0})
dfbp2.valet = dfbp2.valet.replace({True: 1, False: 0})
dfbp2.head()
```

```python
dfbp3 = dfbp2.copy(deep = True)
```

```python
dfbp3 = dfbp3[['business_id' ,'garage','street', 'validated', 'lot', 'valet']]
```

```python
#renaming column to merge with larger df
dfbp3.rename(columns={'business_id':'business_id2'}, inplace=True)
```

```python
dfbp3.head()
```

```python
df2 = pd.merge(df, dfbp3, left_on='business_id', right_on='business_id2')
```

```python
df2.shape
```

```python
df2.head()
```

```python
#drop business_id2

df2.drop('business_id2', axis=1, inplace=True)
```

```python
df2.shape
```

# Break out Categories

```python
# create isolated categories df

dfcat = df2.copy(deep = True)
```

```python
dfcat = dfcat[['business_id', 'name', 'market', 'categories']]
```

```python
dfcat.head()
```

```python
# split categories into multiple columns, first remove unnecessary words "Restaurants", "Food"


dfcat['categories2'] = dfcat['categories']
```

```python
dfcat['categories2']=dfcat.categories2.str.replace('Restaurants, ', '')
dfcat['categories2'] = dfcat.categories2.str.replace(', Restaurants', '')
dfcat['categories2'] = dfcat.categories2.str.replace('Food, ', '')
```

```python
#check if there are still rows with Restaurants and Food. 

dfcat['categories2'].str.contains(', Food,').sum()
```

```python
dfcat['categories2'].str.contains(', Restaurants,').sum()
```

```python
dfcat.head()
```

```python
# create separate df of categories2 to split apart

dfcat2 = pd.get_dummies(dfcat.categories2.str.split(',').apply(pd.Series).stack()).groupby(level=0).sum()
```

```python
dfcat2.head()
```

```python
dfcat2.shape
```

```python
# remove leading and trailing whitespace because there are duplicate category columns
dfcat2.columns = dfcat2.columns.str.strip()
```

```python
dfcat3 = dfcat2.groupby(by=dfcat2.columns, axis=1).sum()
```

```python
dfcat3.shape
```

# Look at Category Totals to determine top 5% categories to keep because 967 category types is too big and will have sparse data

```python
dfcat3.head()
```

```python
#create isolated df to just look at the total and type

dfcat4 = dfcat3.copy(deep = True)
```

```python
#transpose df
dfcat4 = dfcat4.T
```

```python
dfcat4.head()
```

```python
dfcat4['total']= dfcat4.sum(axis = 1)
```

```python
#new df to isolate total and cuisine types

dfcat5 = dfcat4.copy(deep = True)
```

```python
#move category names into an actual column 'type'
dfcat5['type'] = dfcat5.index 
```

```python
dfcat5.head()
```

```python
dfcat5 = dfcat5[['type', 'total']]
```

```python
#Sort totals to look at top 5%
#Top 5% of 30,930 rows = 1,546. So cuisines above 1,546 will be kept.


dfcat5['total'].sort_values(ascending = [False])
```

I combined Chinese restaurant types and Japanese restaurant types to be able to include Japanese restaurants since they were close to the 1546 threshold

```python
# Create Chinese Total

dfch = dfcat3[(dfcat3['Chinese'] == 1)|(dfcat3['Szechuan']==1)|(dfcat3['Cantonese']==1)| (dfcat3['Taiwanese'] ==1) | dfcat3['Mongolian'] == 1]
```

```python
#Including the other Chinese types will add rows to Chinese

dfch.shape
```

```python
dfcat3['ChineseTotal'] = np.where(((dfcat3['Chinese'] == 1) | (dfcat3['Szechuan'] ==1) | (dfcat3['Cantonese'] ==1) | (dfcat3['Taiwanese'] ==1) | 
                                  (dfcat3['Mongolian'] ==1)),1,0)
```

```python
# Create Japanesetotal to combine other Japanese types to include in larger df

dfj = dfcat3[(dfcat3['Japanese'] == 1)|(dfcat3['Sushi Bars']==1)|(dfcat3['Japanese Curry']==1) | (dfcat3['Izakaya'] ==1) | (dfcat3['Teppanyaki'] ==1)]
dfj.shape
```

```python
dfcat3['JapaneseTotal'] = np.where(((dfcat3['Japanese'] == 1) | (dfcat3['Sushi Bars'] ==1) | (dfcat3['Izakaya'] ==1) | (dfcat3['Japanese Curry'] ==1) | 
                                  (dfcat3['Teppanyaki'] ==1)),1,0)
```
