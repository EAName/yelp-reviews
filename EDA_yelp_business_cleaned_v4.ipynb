```python
# First drop yelp_dataset into your drive (Add shortcut to Drive)
business_json_path = '/content/gdrive/MyDrive/yelp_dataset/yelp_business_clean_version1.csv'
# pandas 2.0 added the string 'None' to the default NA markers. Keep the pandas 1.x defaults so that
# 'None' entries in BusinessParking, Ambience, Music etc. stay as values, as they did when this was first run.
na_markers = ['', '#N/A', '#N/A N/A', '#NA', '-1.#IND', '-1.#QNAN', '-NaN', '-nan', '1.#IND', '1.#QNAN',
              '<NA>', 'N/A', 'NA', 'NULL', 'NaN', 'n/a', 'nan', 'null']
df_b = pd.read_csv(business_json_path, keep_default_na=False, na_values=na_markers)
```

```python
df_columns_deleted = df_b.drop(columns = ['AcceptsInsurance','DietaryRestrictions','ByAppointmentOnly','BusinessAcceptsBitcoin','GoodForMeal','DogsAllowed','RestaurantsTableService','GoodForDancing','CoatCheck','BestNights','DriveThru','Open24Hours','BYOB','Corkage','RestaurantsCounterService'])
```

```python
df_columns_deleted.info()
```

```python
#Two-dimensional count for NaN values in the entire dataframe
df_columns_deleted.isnull().sum().sum()
```

```python
#keep rows with fewer than 10 NaN values (drop rows with 10 or more)
df_droppedNaN = df_columns_deleted[df_columns_deleted.isnull().sum(axis=1) < 10]
```

```python
perc = 7.0
min_count =  int(((100-perc)/100)*df_droppedNaN.shape[0] + 1)
mod_df = df_droppedNaN.dropna( axis=1, 
                thresh=min_count)
print("Modified Dataframe : ")
print(mod_df)
```

```python
mod_df.isnull().sum().sum()
```

```python
# Display only: this result is not assigned, so version 4 keeps its NaNs.
# Missing values are imputed per column downstream (YelpBusinessClean5, Yelp_Linear_Regression_EE).
mod_df.fillna(0)
```

```python
mod_df.replace({False: 0, True: 1}, inplace=True)
```
